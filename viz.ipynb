# Setup

```python
import pickle
import pandas as pd
import pathlib
import matplotlib.pyplot as plt

with open('doc_urls.pickle', 'rb') as f:
    doc_urls = pickle.load(f)
```

# Metro Viz

```python
metros = ['Ann Arbor, MI', 'Detroit, MI']

for description, url in doc_urls.items():
    if url.split('/')[-1].startswith('Metro'):
        path = './data/' + url
        print(f'starting viz for path {path}')
        try:
            df = pd.read_csv(path)
        except Exception as e:
            print(f'failed to read {path}: {e}')
            continue
        try:
            select_df = df[df.RegionName.isin(metros)].set_index('RegionName')
        except AttributeError as e:
            print(f'failed to select {path}: {e}')
            continue

        i = 0
        while True:
            try:
                select_df.iloc[:, i:].T.plot(figsize=(16, 9), xlabel='Date', ylabel=description, title=f'{description} in {metros}')
                break
            except:
                i += 1
                if i > 10:
                    break
```

# Zip Viz

```python
zips = [48154, 48152, 48240, 48336]

for description, url in doc_urls.items():
    url = url.replace('Metro', 'Zip')
    if url.split('/')[-1].startswith('Zip'):
        path = './data/' + url
        print(f'starting viz for path {path}')
        try:
            df = pd.read_csv(path)
        except Exception as e:
            print(f'failed to read {path}: {e}')
            continue
        try:
            select_df = df[df.RegionName.isin(zips)].set_index('RegionName')
        except AttributeError as e:
            print(f'failed to select {path}: {e}')
            continue

        i = 0
        while True:
            try:
                select_df.iloc[:, i:].T.plot(figsize=(16, 9), xlabel='Date', ylabel=description, title=f'{description} in {zips}')
                break
            except:
                i += 1
                if i > 10:
                    break
```
